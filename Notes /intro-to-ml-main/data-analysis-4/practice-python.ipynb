```python
class cl:
    def __init__(self,a,b,c):
        self.a = a
        self.b = b
        self.c = c

    def lrnum(self):
        return f"Larget number is {self.a if self.a >= self.b and self.a >= self.c else (self.b if self.b >= self.c else self.c)}"

c = cl(1,2,3)
print(c.lrnum())
```

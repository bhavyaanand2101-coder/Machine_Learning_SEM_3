```python
# A2

ttd = np.array([[1,2,3,4],[5,6,7,8],[11,22,33,44],[55,66,77,88],[111,222,333,444],[555,666,777,888],[1111,2222,3333,44440],[5555,6666,7777,8888]])
ttd
```

```python
mask = ttd>72
print(mask) # masking is a way to filter out data based on some condition. In this case, the mask will be a boolean array indicating which elements of td are greater than 72.
print(ttd[mask]) # this will print the elements of td that are greater than 72, which in this case will be an empty array since all elements of td are less than 72.
print(ttd[~mask]) # False Values
```

```python
res = np.where(ttd > 100, "Big", "Small") 
res
```

```python
cond = [((ttd > 50) & (ttd < 70)), ((ttd >= 70) & (ttd <= 100))]
choice = ["S", "L"]
np.select(cond, choice, default = "Big")
```

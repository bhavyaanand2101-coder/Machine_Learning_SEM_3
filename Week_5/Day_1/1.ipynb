```python
# Sum of all digits in the existing variable n
n=int(input("Enter a number: "))  # Ensure n is an integer
digit_sum = sum(int(digit) for digit in str(n))
print(f"no is:{n}, sum of all digits in the existing variable n is: {digit_sum}")
```

```python
n = int(input("Enter a number: "))
original_n = n
digit_sum = 0
while n > 0:
    digit_sum += n % 10
    n //= 10
print(f"Number: {original_n}, Sum of digits: {digit_sum}")
```

```python
n=int(input("Enter a number: "))
s=0
while n>0:
    r=n%10
    s=s*10+r
    n=n//10
print(s)
```

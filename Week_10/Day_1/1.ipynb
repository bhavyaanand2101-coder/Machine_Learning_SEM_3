```python
# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


# ============================================
# 2. Load dataset
# ============================================

df = pd.read_csv("CO2 Emissions_Canada.csv")

print("Dataset Shape:")
print(df.shape)

print("\nFirst 5 rows:")
print(df.head())


# ============================================
# 3. Select features and target
# ============================================

X = df[
    [
        "Engine Size(L)",
        "Cylinders",
        "Fuel Consumption City (L/100 km)",
        "Fuel Consumption Hwy (L/100 km)",
        "Fuel Consumption Comb (L/100 km)",
        "Fuel Consumption Comb (mpg)"
    ]
]

y = df["CO2 Emissions(g/km)"]


# ============================================
# 4. Split data into training and testing
# ============================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data:", X_test.shape)


# ============================================
# 5. Feature Scaling
# ============================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ============================================
# 6. Create KNN model
# ============================================

knn = KNeighborsRegressor(
    n_neighbors=5
)


# ============================================
# 7. Train model
# ============================================

knn.fit(X_train_scaled, y_train)


# ============================================
# 8. Make predictions
# ============================================

y_pred = knn.predict(X_test_scaled)


# ============================================
# 9. Evaluate model
# ============================================

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)


print("\n========== MODEL PERFORMANCE ==========")

print("MAE :", mae)

print("MSE :", mse)

print("RMSE:", rmse)

print("R2 Score:", r2)


# ============================================
# 10. Compare actual vs predicted
# ============================================

result = pd.DataFrame({
    "Actual CO2": y_test.values,
    "Predicted CO2": y_pred
})

print("\nActual vs Predicted:")
print(result.head(10))


# ============================================
# 11. Find best K value
# ============================================

k_values = range(1, 21)

errors = []

for k in k_values:

    model = KNeighborsRegressor(
        n_neighbors=k
    )

    model.fit(X_train_scaled, y_train)

    prediction = model.predict(X_test_scaled)

    rmse_value = np.sqrt(
        mean_squared_error(y_test, prediction)
    )

    errors.append(rmse_value)


# ============================================
# 12. Plot K vs RMSE
# ============================================

plt.figure(figsize=(8, 5))

plt.plot(k_values, errors, marker="o")

plt.xlabel("K Value")

plt.ylabel("RMSE")

plt.title("KNN: K Value vs RMSE")

plt.xticks(k_values)

plt.grid()

plt.show()


# ============================================
# 13. Find best K
# ============================================

best_k = k_values[np.argmin(errors)]

print("\nBest K:", best_k)

print("Lowest RMSE:", min(errors))


# ============================================
# 14. Train final model using best K
# ============================================

final_model = KNeighborsRegressor(
    n_neighbors=best_k
)

final_model.fit(
    X_train_scaled,
    y_train
)

final_prediction = final_model.predict(
    X_test_scaled
)


# ============================================
# 15. Final performance
# ============================================

final_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        final_prediction
    )
)

final_mae = mean_absolute_error(
    y_test,
    final_prediction
)

final_r2 = r2_score(
    y_test,
    final_prediction
)

print("\n========== FINAL MODEL ==========")

print("Best K:", best_k)

print("MAE:", final_mae)

print("RMSE:", final_rmse)

print("R2 Score:", final_r2)
```

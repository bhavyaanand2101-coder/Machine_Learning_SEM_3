```python
# te a function to find the distnacve between 2 poiunts 
import math

def calculate_distance(p1: tuple, p2: tuple) -> float:
    x1, y1 = p1
    x2, y2 = p2
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Example usage:
print(calculate_distance((1, 2), (4, 6)))  # Output: 5.0
```

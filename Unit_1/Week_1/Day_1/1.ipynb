```python
n=[{"Name":"Bhavya","Age":20},
   {"Name":"Rohit","Age":21},
   {"Name":"Rohit","Age":19},
   {"Name":"Rohit","Age":22}]
print(sorted(n,key=lambda x:x["Age"]))
print(sorted(n,key=lambda x:x["Age"],reverse=True))
print(sorted(n)) #Dictionary is not comparable, so it will throw an error
```

```python
n=[("Bhavya",20),
   ("Rohit",21),
   ("Priyanshu",19),
   ("Pawan",22)]
print(sorted(n,key=lambda x:x[1]))
print(sorted(n,key=lambda x:x[1],reverse=True))
print(sorted(n)) #Tuple is comparable, so it will not throw an error
```

```python
n=[{"Bhavya":20,"sal":10000},
   {"Rohit":21,"sal":20000},
   {"Priyanshu":19,"sal":30000},
   {"Pawan":22,"sal":40000}]
print(sorted(n,key=lambda x:x["sal"],reverse=True))
```

```python
n=[12,34,45,87,64,68,99,100,200,300,45,76,8765,12,12,55,55,55,55,55,55,55,55,55,55,55,55,55,55]
#Frequency of each element in the list
print(len(set(n))) #Number of unique elements in the list
print(set(n)) #Unique elements in the list
d={}
for i in n:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d) #Frequency of each element in the list
```

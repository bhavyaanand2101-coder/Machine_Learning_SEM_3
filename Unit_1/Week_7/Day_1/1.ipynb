```python
import numpy as np

# 1. Create a 3D array with 9 elements
array3d = np.arange(1, 10).reshape(1, 3, 3)
print("3D array:\n", array3d)

# 2. Create a 2D array with 32 elements
array_2d = np.arange(1, 33).reshape(4, 8)
print("\n2D array:\n", array_2d)

# 3. Reshape the 2D array into two different shapes
reshaped_1 = array_2d.reshape(8, 4)
reshaped_2 = array_2d.reshape(2, 16)

print("\nReshaped to (8, 4):\n", reshaped_1)
print("\nReshaped to (2, 16):\n", reshaped_2)

# 4. Find the transpose
transposed_array = array_2d.T
print("\nTranspose:\n", transposed_array)
```
